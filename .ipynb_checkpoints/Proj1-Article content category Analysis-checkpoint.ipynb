### Visualization for (1.2) using plotly

```python
import plotly.express as px

# Filter years
filtered = df[(df["last_updated_year"] >= 2010) &
              (df["last_updated_year"] <= 2025)]

# Top 15 tags overall (2010-2025)
top_tags = (
    filtered["clean_tags"]
    .value_counts()
    .head(15)
    .index
)

# Count occurrences of those tags per year
plot_df = (
    filtered[filtered["clean_tags"].isin(top_tags)]
    .groupby(["last_updated_year", "clean_tags"])
    .size()
    .reset_index(name="Count")
)

fig = px.bar(
    plot_df,
    x="last_updated_year",
    y="Count",
    color="clean_tags",
    barmode="group",
    text="Count",
    title="Top 15 Tags by Year (2020–2025)",
    color_discrete_sequence=px.colors.qualitative.G10  # GFG-like color palette
)

fig.update_traces(
    textposition="outside",
    marker_line_color="white",
    marker_line_width=0.8
)

fig.update_layout(
    paper_bgcolor="black",
    plot_bgcolor="black",
    template="plotly_dark",
    width=1400,
    height=700,

    title=dict(
        text="Top 15 Tags by Year (2020–2025)",
        x=0.5,
        font=dict(size=22, color="white")
    ),

    font=dict(color="white", size=14),

    xaxis=dict(
        title="Year",
        tickmode="linear",
        showgrid=False,
        color="white"
    ),

    yaxis=dict(
        title="Count",
        gridcolor="gray",
        color="white"
    ),

    legend=dict(
        title="Categories",
        bgcolor="black",
        font=dict(color="white")
    )
)

fig.show()
```

### Visualization for (1.3) using plotly

```python
import plotly.express as px

# Top 10 categories
top10 = (
    df["clean_tags"]
    .value_counts()
    .head(10)
    .reset_index()
)

top10.columns = ["Category", "Count"]

# Percentage of overall dataset
total = len(df)
top10["Percentage"] = (top10["Count"] / total * 100).round(2)

# GFG Green Palette
gfg_green = [
    "#2F8D46", "#3FA55B", "#56B870", "#6BCB86", "#81D99C",
    "#1F6F3D", "#2C7A4B", "#3D8F5A", "#4FA368", "#66B97D"
]

fig = px.pie(
    top10,
    names="Category",
    values="Percentage",
    hole=0.15,
    color="Category",
    color_discrete_sequence=gfg_green,
    width=1400,
    height=900,
    title="Top 10 Categories (% of Overall Dataset)"
)

fig.update_traces(
    textposition="outside",
    textinfo="label+percent",
    textfont_size=18,
    pull=[0.03] * len(top10),
    hovertemplate="<b>%{label}</b><br>Count: %{customdata[0]}<br>Percentage: %{value:.2f}%<extra></extra>",
    customdata=top10[["Count"]]
)

fig.update_layout(
    paper_bgcolor="black",
    plot_bgcolor="black",
    template="plotly_dark",
    font=dict(color="white", size=16),
    title=dict(
        x=0.5,
        font=dict(size=28, color="white")
    ),
    legend=dict(
        bgcolor="black",
        font=dict(color="white", size=14)
    ),
    margin=dict(t=80, l=50, r=250, b=50)
)

fig.show()
```

### Visualization for (1.5)

```python
import pandas as pd
import plotly.express as px

lst = []

# Collect top 10 tags from each year
for year in range(2020, 2026):
    lst += list(
        df[df["last_updated_year"] == year]["clean_tags"]
        .value_counts()
        .head(10)
        .index
    )

# Count how many years each tag appeared in the Top 10
tag_count = {}
for tag in lst:
    if tag in tag_count:
        tag_count[tag] += 1
    else:
        tag_count[tag] = 1

# Convert to DataFrame
plot_df = pd.DataFrame(
    tag_count.items(),
    columns=["Category", "Years_Appeared"]
).sort_values("Years_Appeared", ascending=False)

fig = px.bar(
    plot_df,
    x="Category",
    y="Years_Appeared",
    text="Years_Appeared",
    color_discrete_sequence=["#2F8D46"],   # GeeksforGeeks Green
    title="Categories Consistently Appearing in Top 10 (2020–2025)"
)

fig.update_traces(
    textposition="outside",
    marker_line_color="white",
    marker_line_width=1
)

fig.update_layout(
    paper_bgcolor="black",      # Outside background
    plot_bgcolor="black",       # Plot background
    font=dict(color="white", size=14),
    title_font=dict(size=20, color="white"),
    xaxis=dict(
        title="Category",
        tickangle=-45,
        color="white",
        showgrid=False
    ),
    yaxis=dict(
        title="Number of Years in Top 10",
        color="white",
        gridcolor="gray"
    ),
    template="plotly_dark",
    width=1200,
    height=600
)

fig.show()
```

### 1.6 Do certain cateogory contents tend to have more images than others?

```python
df.groupby('clean_tags')["no_of_images"].mean().sort_values(ascending=False).head(20)  #These are top 20 categories have more images than others
```

```python
import plotly.graph_objects as go
import pandas as pd

# Average images per category
plot_df = (
    df.groupby("clean_tags")
      .agg(
          avg_images=("no_of_images", "mean"),
          app_count=("clean_tags", "size")
      )
      .sort_values("avg_images", ascending=False)
      .reset_index()
)

thresholds = list(range(100, int(plot_df["app_count"].max()) + 100, 100))

fig = go.Figure()

# Create one trace for each threshold
for t in thresholds:

    filtered = plot_df[plot_df["app_count"] >= t]

    fig.add_trace(
        go.Scatter(
            x=filtered["clean_tags"],
            y=filtered["avg_images"],
            mode="lines+markers",
            line=dict(color="#2F8D46", width=3),
            marker=dict(
                size=7,
                color="#2F8D46",
                line=dict(color="#7CFC98", width=2)
            ),
            visible=False,
            hovertemplate="<b>%{x}</b><br>Average Images: %{y:.2f}<extra></extra>"
        )
    )

# Make first trace visible
fig.data[0].visible = True

# Slider
sliders = [dict(
    active=0,
    currentvalue={
        "prefix": "Min Total Count: ",
        "font": {"size": 14, "color": "white"},
        "xanchor": "right"
    },
    steps=steps,

    # Position
    x=0.75,          # Move to the right (0 to 1)
    y=1.18,          # Move above the plot
    len=0.22,        # Slider length

    xanchor="left",
    yanchor="top",

    pad={"t": 10, "b": 10},

    bgcolor="#111217",
    bordercolor="#2F8D46",
    borderwidth=2
)]

fig.update_layout(

    sliders=sliders,

    template="plotly_dark",
    paper_bgcolor="#111217",
    plot_bgcolor="#111217",

    width=1500,
    height=700,

    title="Average Images per Tag",

    font=dict(color="white"),

    xaxis=dict(
        title="Category",
        tickangle=-45,
        showgrid=True,
        gridcolor="#22252E"
    ),

    yaxis=dict(
        title="Average Images",
        showgrid=True,
        gridcolor="#22252E"
    ),

    hovermode="x unified"
)

fig.show()
```
