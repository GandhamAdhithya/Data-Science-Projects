### 4.2) Top 15 category of Book count?

```python
df["category"].value_counts().head(15)
```

```python
df["category"].value_counts().head(15).plot(kind="pie",autopct="%1.f%%")
```

```python
import plotly.graph_objects as go

data = df["category"].value_counts().head(15)

colors = [
    "#E3F2FD","#BBDEFB","#90CAF9","#64B5F6","#42A5F5",
    "#2196F3","#1E88E5","#1976D2","#1565C0","#0D47A1",
    "#82B1FF","#448AFF","#2979FF","#2962FF","#0039CB"
]

fig = go.Figure(go.Pie(
    labels=data.index,
    values=data.values,
    hole=0.4,
    textposition="outside",          # Labels outside
    textinfo="label+percent",
    textfont=dict(size=16, color="black"),
    automargin=True,
    marker=dict(
        colors=colors,
        line=dict(color="white", width=3)
    ),
    pull=[0.03]*len(data),
    hovertemplate="<b>%{label}</b><br>Count: %{value}<br>%{percent}<extra></extra>"
))

fig.update_layout(
    title=dict(
        text="<b>Top 15 Categories</b>",
        x=0.5,
        font=dict(size=28, color="#0D47A1")
    ),
    paper_bgcolor="white",
    plot_bgcolor="white",
    width=1100,      # Bigger chart
    height=850,      # Bigger chart
    margin=dict(t=80, l=80, r=80, b=80),
    font=dict(size=16),
    showlegend=False
)

fig.show()
```

### 4.3) Top 15 Authors by number of books published?

```python
df["author_id"].value_counts().head(15).plot(kind="bar")
```

```python
import plotly.graph_objects as go

# Top 15 authors
data = df["author_id"].value_counts().head(15)

# Percentage of the displayed top 15
percent = data / data.sum() * 100

colors = [
    "#E3F2FD","#D6ECFF","#BBDEFB","#90CAF9","#64B5F6",
    "#42A5F5","#2196F3","#1E88E5","#1976D2","#1565C0",
    "#0D47A1","#2962FF","#1A73E8","#0039CB","#001F8F"
]

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=data.index.astype(str),
        y=data.values,
        text=[f"{p:.1f}%" for p in percent],
        textposition="outside",
        textfont=dict(size=15, color="#0D47A1"),
        marker=dict(
            color=colors,
            line=dict(color="white", width=2)
        ),
        hovertemplate="<b>Author ID:</b> %{x}<br><b>Count:</b> %{y}<br><b>Percentage:</b> %{text}<extra></extra>"
    )
)

fig.update_layout(
    title=dict(
        text="<b>Top 15 Author IDs</b>",
        x=0.5,
        font=dict(size=28, color="#0D47A1")
    ),
    paper_bgcolor="white",
    plot_bgcolor="white",
    width=1200,
    height=700,
    xaxis=dict(
        title="Author ID",
        tickfont=dict(size=13),
        showgrid=False
    ),
    yaxis=dict(
        title="Count",
        gridcolor="#E5E7EB",
        zeroline=False
    ),
    font=dict(
        family="Arial",
        size=15,
        color="black"
    ),
    bargap=0.25
)

# Rounded bars (Plotly 5.19+)
fig.update_traces(marker_line_width=2, marker_cornerradius=8)

fig.show()
```

### 4.4) Yearly Trend in Books Published(Earlier->Latest)?

```python
import plotly.graph_objects as go

# Books published each year
data = df["publish_year"].value_counts().sort_index()

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=data.index,
        y=data.values,
        mode="lines+markers+text",
        text=data.values,
        textposition="top center",
        textfont=dict(size=12, color="#0D47A1"),
        line=dict(
            color="#1565C0",
            width=5,
            shape="spline",
            smoothing=1.2
        ),
        marker=dict(
            size=11,
            color="#2196F3",
            line=dict(color="white", width=2)
        ),
        hovertemplate="<b>Year:</b> %{x}<br><b>Books:</b> %{y}<extra></extra>",
        fill="tozeroy",
        fillcolor="rgba(33,150,243,0.15)"
    )
)

fig.update_layout(
    title=dict(
        text="<b>Books Published by Year</b>",
        x=0.5,
        font=dict(size=28, color="#0D47A1")
    ),
    paper_bgcolor="white",
    plot_bgcolor="white",
    width=1300,
    height=700,
    hovermode="x unified",
    font=dict(
        family="Arial",
        size=15,
        color="black"
    ),
    hoverlabel=dict(
        bgcolor="#0D47A1",
        font_size=15,
        font_color="white"
    ),
    xaxis=dict(
        title="Publication Year",
        showgrid=False,
        showline=True,
        linewidth=2,
        linecolor="#1565C0",
        tickangle=-45
    ),
    yaxis=dict(
        title="Number of Books",
        gridcolor="#E8EEF7",
        zeroline=False,
        showline=True,
        linewidth=2,
        linecolor="#1565C0"
    )
)

fig.show()
```

### 4.6) Top 15 Authors by total pages Written?

```python
df_=df[df["pages"]<5000] # Here we modify to each book conssists max 5000 pages or below that books we considering
df_.groupby("author_id")["pages"].sum().sort_values(ascending=False).head(15)
```

```python
import plotly.graph_objects as go

# Filter books with <= 5000 pages
df_ = df[df["pages"] < 5000]

# Top 15 authors by total pages
data = (
    df_
    .groupby("author_id")["pages"]
    .sum()
    .sort_values(ascending=False)
    .head(15)
)

# Percentage of total pages among Top 15
percent = data / data.sum() * 100

colors = [
    "#E3F2FD","#D6ECFF","#BBDEFB","#90CAF9","#64B5F6",
    "#42A5F5","#2196F3","#1E88E5","#1976D2","#1565C0",
    "#0D47A1","#2962FF","#1A73E8","#0039CB","#001F8F"
]

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=data.index.astype(str),
        y=data.values,
        text=[f"{p:.1f}%" for p in percent],
        textposition="outside",
        textfont=dict(size=14, color="#0D47A1"),
        marker=dict(
            color=colors,
            line=dict(color="white", width=2)
        ),
        hovertemplate=(
            "<b>Author ID:</b> %{x}<br>"
            "<b>Total Pages:</b> %{y:,}<br>"
            "<b>Percentage:</b> %{text}<extra></extra>"
        )
    )
)

fig.update_layout(
    title=dict(
        text="<b>Top 15 Authors by Total Pages</b>",
        x=0.5,
        font=dict(size=28, color="#0D47A1")
    ),
    paper_bgcolor="white",
    plot_bgcolor="white",
    width=1400,
    height=750,
    bargap=0.25,
    font=dict(
        family="Arial",
        size=15,
        color="black"
    ),
    xaxis=dict(
        title="Author ID",
        showgrid=False,
        showline=True,
        linewidth=2,
        linecolor="#1565C0"
    ),
    yaxis=dict(
        title="Total Pages",
        gridcolor="#E8EEF7",
        zeroline=False,
        showline=True,
        linewidth=2,
        linecolor="#1565C0"
    ),
    hoverlabel=dict(
        bgcolor="#0D47A1",
        font_size=15,
        font_color="white"
    )
)

# Rounded bars (Plotly >= 5.19)
fig.update_traces(
    marker_line_width=2,
    marker_cornerradius=8
)

fig.show()
```
