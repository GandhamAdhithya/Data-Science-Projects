```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\Users\Adhithya\OneDrive\문서\Desktop\Data Science Projects\List of Countries by Sugarcane Production.csv")
df.head()
```

```python
df.shape
```

## 1.Data Cleaning

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df["Production (Tons)"]=df["Production (Tons)"].str.replace(".","")
```

```python
df["Production per Person (Kg)"]=df["Production per Person (Kg)"].str.replace(".","").str.replace(",","")
```

```python
df["Acreage (Hectare)"]=df["Acreage (Hectare)"].str.replace(".","")
```

```python
df["Yield (Kg / Hectare)"]=df["Yield (Kg / Hectare)"].str.replace(".","").str.replace(",",".")
```

```python
df.head()
```

```python
df.rename(columns={"Production (Tons)":"Production(Tons)"},inplace=True)
df.rename(columns={"Production per Person (Kg)":"Production_per_Person(Kg)"},inplace=True)
df.rename(columns={"Acreage (Hectare)":"Acreage(Hectare)"},inplace=True)
df.rename(columns={"Yield (Kg / Hectare)":"Yield(Kg/Hectare)"},inplace=True)
```

```python
df.head()
```

```python
del df["Unnamed: 0"]
```

```python
df.head()
```

```python
df.dtypes
```

```python
df["Production(Tons)"]          =df["Production(Tons)"].astype(float)
df["Production_per_Person(Kg)"] =df["Production_per_Person(Kg)"].astype(float)
df["Acreage(Hectare)"]          =df["Acreage(Hectare)"].astype(float)
df["Yield(Kg/Hectare)"]          =df["Yield(Kg/Hectare)"].astype(float)
```

```python
df.dtypes
```

## Analysis For Continents

```python
df_continent=df.groupby("Continent").sum()
df_continent
```

### Which continent produce more SugarCane?

```python
df_continent["Production(Tons)"].sort_values(ascending=False)
```

```python
df_continent["Production(Tons)"].sort_values(ascending=False).plot(kind="bar")
```

### Do number of countries in a continent Effects production of SugarCane?

```python
df["Continent"].value_counts()  # Answer is "No", number of countries in a continent not Effects production of SugarCane
```

```python
df_continent["No_of_countries"]=df.groupby("Continent").count()["Country"]
```

```python
df_continent
```

```python
continent_names=df_continent.index.to_list()
```

```python
continent_names
```

```python
sns.lineplot(data=df_continent,x="No_of_countries",y="Production(Tons)")
plt.xticks(df_continent["No_of_countries"],continent_names,rotation=90)
plt.show()
```
