### 2.1)Cleaning names,duration & Votes

```python
df["name"]=df["name"].str.strip() 
df["duration"]=df["duration"].str.replace(" min","").str.replace(',','').astype(int)
df["votes"]=df["votes"].str.replace(",",'').astype(int)
```

### 2.2) Cleaning Year column

```python
import re
start_year=[]
end_year=[]
for text in df['year']:
    start_year.append(text)
    end_year.append(text)
df['start_year']=start_year
df['end_year']=end_year
    
```

### 2.3) cleaning director_id,director_names

```python
director_ids_list=[]
director_names_list=[]
for director_ids,director_names in df[['directors_id','directors_name']].values:
    director_ids_list.append(director_ids)
    director_names_list.append(director_names)
df['directors_id']=director_ids_list
df['directors_name']=director_names_list
directors_dict=dict(zip(director_ids_list,director_names_list))
with open('directors.json','w') as f:
    f.write(json.dumps(directors_dict))
del director_ids_list
del director_names_list
del df['directors_name']       #here no need of director_names column because we stored both in dict as json file if we search with director_id in dict automatically it gives director_name for that id
```

### 2.4)Mapping starnames and Id's

```python
star_ids_list=[]
star_names_list=[]
for star_ids,star_names in df[['stars_id','stars_name']].values:
    star_ids_list.append(star_ids)
    star_names_list.append(star_names)
df['stars_id']=star_ids_list
df['stars_name']=star_names_list
star_dict=dict(zip(star_ids_list,star_names_list))
with open('star.json','w') as f:
    f.write(json.dumps(star_dict))
del star_ids_list
del star_names_list
del df['stars_name']   
```

```python
import matplotlib.pyplot as plt
import mplcursors

# Prepare data
dist = (
    df_.groupby(["certificate_category", "certificate"])
       .size()
       .reset_index(name="count")
)

dist = dist.sort_values(
    ["certificate_category", "count"],
    ascending=[True, False]
)

# Theme
BG = "#0D1117"
TITLE = "#F5C518"

colors = [
    "#F5C518", "#58A6FF", "#3FB950",
    "#FF7B72", "#D2A8FF", "#FFA657",
    "#79C0FF", "#8B949E"
]

cats = dist["certificate_category"].unique()

cols = 2
rows = (len(cats) + cols - 1) // cols

fig, axes = plt.subplots(
    rows,
    cols,
    figsize=(14, rows * 6),
    facecolor=BG
)

axes = axes.flatten()

for ax in axes:
    ax.set_facecolor(BG)

for i, cat in enumerate(cats):

    temp = dist[dist["certificate_category"] == cat].reset_index(drop=True)

    ax = axes[i]

    wedges, texts, autotexts = ax.pie(
        temp["count"],
        labels=None,
        colors=colors[:len(temp)],
        startangle=90,
        autopct="%1.1f%%",
        pctdistance=0.78,
        wedgeprops=dict(
            width=0.40,
            edgecolor=BG,
            linewidth=2
        )
    )

    # Donut hole
    centre = plt.Circle((0, 0), 0.60, fc=BG)
    ax.add_artist(centre)

    # Category name
    ax.text(
        0, 0,
        cat,
        ha="center",
        va="center",
        fontsize=13,
        color=TITLE,
        fontweight="bold"
    )

    # Legend
    ax.legend(
        temp["certificate"],
        loc="lower center",
        bbox_to_anchor=(0.5, -0.22),
        fontsize=8,
        frameon=False,
        labelcolor="white",
        ncol=2
    )

    # Interactive Hover
    cursor = mplcursors.cursor(wedges, hover=True)

    @cursor.connect("add")
    def on_add(sel, temp=temp):
        idx = sel.index
        total = temp["count"].sum()
        value = temp.iloc[idx]["count"]
        pct = value / total * 100

        sel.annotation.set(
            text=f"""Certificate : {temp.iloc[idx]['certificate']}
Count       : {value:,}
Percentage  : {pct:.2f}%"""
        )

        sel.annotation.get_bbox_patch().set(
            facecolor="#1E1E1E",
            edgecolor="#F5C518",
            alpha=0.95
        )

# Remove empty axes
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.suptitle(
    "IMDb Certificate Distribution Dashboard",
    fontsize=24,
    color=TITLE,
    fontweight="bold"
)

plt.tight_layout(rect=[0, 0, 1, 0.95])

plt.show()
```

```python
import plotly.subplots as sp
import plotly.graph_objects as go

# -------- Filter only Kids, Family and Adult --------
filtered = df_[df_["certificate_category"].isin(["Kids", "Family", "Adult"])]

# -------- Aggregate Votes --------
dist = (
    filtered
    .groupby(["certificate_category", "certificate"])["votes"]
    .sum()
    .reset_index()
)

dist = dist.sort_values(
    ["certificate_category", "votes"],
    ascending=[True, False]
)

cats = ["Kids", "Family", "Adult"]   # Display order

# -------- Dashboard Layout --------
cols = 2
rows = (len(cats) + cols - 1) // cols

fig = sp.make_subplots(
    rows=rows,
    cols=cols,
    specs=[[{"type": "domain"}] * cols for _ in range(rows)],
    subplot_titles=cats
)

# IMDb-inspired colors
colors = [
    "#F5C518", "#58A6FF", "#3FB950",
    "#FF7B72", "#D2A8FF", "#FFA657",
    "#79C0FF", "#A5D6A7", "#FF9F1C",
    "#00C2A8"
]

# -------- Create Donut Charts --------
for i, cat in enumerate(cats):

    temp = dist[dist["certificate_category"] == cat]

    r = i // cols + 1
    c = i % cols + 1

    fig.add_trace(
        go.Pie(
            labels=temp["certificate"],
            values=temp["votes"],
            hole=0.60,
            textinfo="percent",
            textfont=dict(size=12),
            marker=dict(
                colors=colors,
                line=dict(color="white", width=2)
            ),
            pull=[0.03] * len(temp),
            hovertemplate=
                "<b>%{label}</b><br>"
                "Votes: %{value:,}<br>"
                "Share: %{percent}<extra></extra>",
            showlegend=False
        ),
        row=r,
        col=c
    )

# -------- Dashboard Style --------
fig.update_layout(
    template="plotly_dark",
    paper_bgcolor="#0D1117",
    plot_bgcolor="#0D1117",
    title=dict(
        text="<b>IMDb Certificate Votes Dashboard</b>",
        x=0.5,
        font=dict(size=26, color="#F5C518")
    ),
    font=dict(color="white", family="Arial"),
    height=350 * rows,
    margin=dict(t=90, l=20, r=20, b=20)
)

fig.show(config={
    "displaylogo": False,
    "scrollZoom": True,
    "displayModeBar": True
})
```
