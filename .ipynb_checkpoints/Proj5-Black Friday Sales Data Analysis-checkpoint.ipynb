# 3.Analyzing Gender

```python
df[df["Gender"]=="M"]
```

### How many males and Females are purchasing products

```python
len(df[df["Gender"]=="M"]),len(df[df["Gender"]=="F"])
```

```python
data=pd.DataFrame({'Ratio' :[len(df[df["Gender"]=="M"]),len(df[df["Gender"]=="F"])]},index=["Male","Female"])
```

```python
data.plot.pie(y="Ratio",autopct="%1.f%%",figsize=(5,5))
```

```python
#Another method 
df.groupby("Gender").size().plot(kind="pie",autopct="%1.f%%",title="Gender Ratio(%)")
```

### How much total cost spend by male and female to purchase products?

```python
df.groupby("Gender")["Purchase"].sum()
```

```python
df.groupby("Gender")["Purchase"].sum().plot(kind="pie",autopct="%1.f%%")
```

```python
df.groupby("Gender")["Purchase"].mean().plot(kind="pie",autopct="%1.2f%%")
```
